# Phase 7 — Model Training & Comparison Notebook

**Project:** AI-Powered Customer Feedback Intelligence System  
**Goal:** Train and compare three baseline machine learning algorithms on 20,000 TF-IDF features.

### Classifiers Benchmark:
1. **Multinomial Naive Bayes** (Fast probabilistic baseline)
2. **Logistic Regression** (Linear probabilistic model with L2 regularization)
3. **Linear SVM (`LinearSVC`)** (Maximum margin hyperplane separator)

```python
import pandas as pd
import numpy as np
import joblib
import time
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

# Load pre-fitted TF-IDF vectorizer and datasets
df = pd.read_csv('../data/processed/reviews_cleaned.csv')
df['clean_text'] = df['clean_text'].fillna('')
tfidf = joblib.load('../models/tfidf_vectorizer.pkl')

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    df['clean_text'], df['Sentiment'], test_size=0.20, random_state=42, stratify=df['Sentiment']
)

X_train_tfidf = tfidf.transform(X_train_raw)
X_test_tfidf = tfidf.transform(X_test_raw)

print(f'Train matrix: {X_train_tfidf.shape}')
print(f'Test matrix : {X_test_tfidf.shape}')
```

## 1. Multinomial Naive Bayes
Applies Bayes' Theorem assuming independence among TF-IDF feature frequencies.

```python
nb = MultinomialNB()
start = time.time()
nb.fit(X_train_tfidf, y_train)
nb_time = time.time() - start
nb_acc = accuracy_score(y_test, nb.predict(X_test_tfidf)) * 100
print(f'Naive Bayes Accuracy: {nb_acc:.2f}% ({nb_time:.2f}s)')
```

## 2. Logistic Regression
Optimizes log-loss over high-dimensional sparse TF-IDF features.

```python
lr = LogisticRegression(max_iter=1000, random_state=42)
start = time.time()
lr.fit(X_train_tfidf, y_train)
lr_time = time.time() - start
lr_acc = accuracy_score(y_test, lr.predict(X_test_tfidf)) * 100
print(f'Logistic Regression Accuracy: {lr_acc:.2f}% ({lr_time:.2f}s)')
```

## 3. Linear SVM (LinearSVC)
Finds the maximum margin hyper-plane separating Positive, Neutral, and Negative clusters.

```python
svm = LinearSVC(max_iter=2000, random_state=42)
start = time.time()
svm.fit(X_train_tfidf, y_train)
svm_time = time.time() - start
svm_acc = accuracy_score(y_test, svm.predict(X_test_tfidf)) * 100
print(f'Linear SVM Accuracy: {svm_acc:.2f}% ({svm_time:.2f}s)')
```
