# Phase 8 — Model Evaluation & Metrics Notebook

**Project:** AI-Powered Customer Feedback Intelligence System  
**Goal:** Perform rigorous multi-metric evaluation across all 3 trained models (Accuracy, Precision, Recall, F1, Macro F1, Confusion Matrices).

### Key Metrics Covered:
* **Accuracy**: Overall fraction of correct predictions.
* **Precision**: Fraction of predicted positive instances that are actually positive.
* **Recall**: Fraction of actual positive instances correctly captured.
* **Macro F1-Score**: Equal-weighted arithmetic mean of F1 scores across Positive, Neutral, and Negative classes.

```python
import pandas as pd
import numpy as np
import joblib
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

# Load pre-trained models and test features
df = pd.read_csv('../data/processed/reviews_cleaned.csv')
df['clean_text'] = df['clean_text'].fillna('')
tfidf = joblib.load('../models/tfidf_vectorizer.pkl')
lr_model = joblib.load('../models/logistic_regression_model.pkl')

_, X_test_raw, _, y_test = train_test_split(
    df['clean_text'], df['Sentiment'], test_size=0.20, random_state=42, stratify=df['Sentiment']
)
X_test_tfidf = tfidf.transform(X_test_raw)

print(f'Test matrix shape: {X_test_tfidf.shape}')
```

## Classification Report — Logistic Regression

```python
y_pred = lr_model.predict(X_test_tfidf)
labels = ['Negative', 'Neutral', 'Positive']
print(classification_report(y_test, y_pred, labels=labels, digits=4))
print('Confusion Matrix:')
print(pd.DataFrame(confusion_matrix(y_test, y_pred, labels=labels), index=labels, columns=labels))
```
